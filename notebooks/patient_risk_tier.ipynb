```python
import pandas as pd
import numpy as np
```

```python
demo_df=pd.read_csv("patient_demographics.csv")
demo_df
```

```python
df=demo_df.copy(deep=True)
df
```

```python
main_df=pd.read_csv('feature_engineered_df.csv')
main_df
```

```python
df=df[df['patient_id'].isin(main_df['patient_id'])]
df
```

```python
df['age_band'].value_counts()
```

```python
df=pd.get_dummies(df,columns=['age_band'],drop_first=True,dtype=int)
```

```python
df['insurance_type'].value_counts()
```

```python
df=pd.get_dummies(df,columns=['insurance_type'],drop_first=True,dtype=int)
```

```python
df
```

```python
df=pd.get_dummies(df,columns=['gender'],drop_first=True,dtype=int)
df
```

```python
df['state'].value_counts()
```

```python
df=pd.get_dummies(df,columns=['state'],drop_first=True,dtype=int)
df
```

```python
df
```

```python
df.info()
```

```python
df=df.merge(main_df,on='patient_id',how='inner')
df
```
